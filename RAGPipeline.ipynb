```python
# =============================================================================
# STAGE 1 — CONTEXT-AWARE CHUNKING  (DocumentIngestionEngine)
# =============================================================================

class DocumentIngestionEngine:
    """
    Loads PDF or Markdown files and produces header-aware, size-bounded
    Document chunks that carry full section breadcrumb metadata.

    Two-pass split strategy:
      Pass 1 — MarkdownHeaderTextSplitter  → semantic structural boundaries
      Pass 2 — RecursiveCharacterTextSplitter → hard token-budget enforcement

    Every sub-chunk receives a 'section_path' breadcrumb injected into
    both its metadata and the beginning of its text, so the LLM always
    knows WHERE in the document each passage originates (solves
    "lost-in-the-middle" context neglect).
    """

    MARKDOWN_HEADERS = [
        ("#",    "header_1"),
        ("##",   "header_2"),
        ("###",  "header_3"),
        ("####", "header_4"),
    ]

    RECURSIVE_SEPARATORS = [
        "\n\n\n", "\n\n", "\n", ". ", "? ", "! ", "; ", " ", ""
    ]

    def __init__(self, chunk_size: int = 512, chunk_overlap: int = 64):
        self.chunk_size    = chunk_size
        self.chunk_overlap = chunk_overlap

        self._md_splitter = MarkdownHeaderTextSplitter(
            headers_to_split_on=self.MARKDOWN_HEADERS,
            strip_headers=False,       # keep header text inside chunk body
            return_each_line=False,
        )
        self._recursive_splitter = RecursiveCharacterTextSplitter(
            separators=self.RECURSIVE_SEPARATORS,
            chunk_size=self.chunk_size,
            chunk_overlap=self.chunk_overlap,
            length_function=len,
            add_start_index=True,
        )

    # ── Public entry point ────────────────────────────────────────────────────

    def ingest(self, file_path: str) -> List[Document]:
        path = Path(file_path)
        if not path.exists():
            raise FileNotFoundError(f"Document not found: {file_path}")

        suffix = path.suffix.lower()
        logger.info("Loading document: %s  (type=%s)", path.name, suffix)

        if suffix == ".pdf":
            raw_docs = self._load_pdf(path)
        elif suffix in {".md", ".markdown"}:
            raw_docs = self._load_markdown(path)
        else:
            raise ValueError(
                f"Unsupported file type '{suffix}'. "
                "Supported formats: .pdf  .md  .markdown"
            )

        chunks = self._split_documents(raw_docs, source_name=path.name)
        logger.info("Produced %d enriched chunks from '%s'.", len(chunks), path.name)
        return chunks

    # ── Private helpers ───────────────────────────────────────────────────────

    def _load_pdf(self, path: Path) -> List[Document]:
        loader = PyPDFLoader(str(path))
        pages  = loader.load()
        for page in pages:
            page.metadata.setdefault("source", path.name)
        return pages

    def _load_markdown(self, path: Path) -> List[Document]:
        loader = UnstructuredMarkdownLoader(str(path), mode="elements")
        docs   = loader.load()
        for doc in docs:
            doc.metadata.setdefault("source", path.name)
        return docs

    def _split_documents(self, docs: List[Document], source_name: str) -> List[Document]:
        all_chunks: List[Document] = []

        for doc in docs:
            # ── Pass 1: semantic header split ─────────────────────────────
            md_chunks = self._md_splitter.split_text(doc.page_content)

            for md_chunk in md_chunks:
                # Merge page-level metadata with header metadata
                merged_meta = {**doc.metadata, **md_chunk.metadata}
                merged_meta["source"] = source_name

                # Build a human-readable breadcrumb for LLM context
                breadcrumb_parts = [
                    merged_meta.get("header_1", ""),
                    merged_meta.get("header_2", ""),
                    merged_meta.get("header_3", ""),
                    merged_meta.get("header_4", ""),
                ]
                breadcrumb = " > ".join(p for p in breadcrumb_parts if p).strip()
                if breadcrumb:
                    merged_meta["section_path"] = breadcrumb

                # ── Pass 2: enforce token budget ──────────────────────────
                sub_docs = self._recursive_splitter.create_documents(
                    texts=[md_chunk.page_content],
                    metadatas=[merged_meta],
                )

                # Prepend breadcrumb into chunk text (LLM always sees context)
                for sub_doc in sub_docs:
                    if breadcrumb and not sub_doc.page_content.startswith("[Section:"):
                        sub_doc.page_content = (
                            f"[Section: {breadcrumb}]\n\n{sub_doc.page_content}"
                        )
                    all_chunks.append(sub_doc)

        return all_chunks

```

```python
# =============================================================================
# STAGE 2 — BGE EMBEDDINGS & PERSISTENT VECTOR STORE  (VectorStoreManager)
# =============================================================================

class VectorStoreManager:
    """
    Manages the Chroma vector store backed by BAAI/bge-large-en-v1.5.
    The local embedding model is entirely free and runs on CPU (or CUDA if
    available). Chroma persists to disk so re-indexing is not needed on restart.
    """

    def __init__(
        self,
        persist_dir: str = CHROMA_PERSIST_DIR,
        collection: str  = COLLECTION_NAME,
    ):
        self.persist_dir = persist_dir
        self.collection  = collection
        self.embeddings  = self._init_embeddings()
        self._store: Optional[Chroma] = None

    @staticmethod
    def _init_embeddings() -> HuggingFaceEmbeddings:
        logger.info("Loading embedding model: %s  (first run downloads ~1.2 GB)", EMBEDDING_MODEL_NAME)
        return HuggingFaceEmbeddings(
            model_name=EMBEDDING_MODEL_NAME,
            model_kwargs={"device": "cpu"},          # change to "cuda" if GPU available
            encode_kwargs={"normalize_embeddings": True},
            query_instruction=(
                "Represent this sentence for searching relevant passages: "
            ),
        )

    # ── Store management ──────────────────────────────────────────────────────

    def add_documents(self, documents: List[Document]) -> None:
        logger.info(
            "Indexing %d chunks into Chroma collection '%s'…",
            len(documents), self.collection,
        )
        if self._store is None:
            self._store = Chroma(
                collection_name=self.collection,
                embedding_function=self.embeddings,
                persist_directory=self.persist_dir,
            )
        self._store.add_documents(documents)
        logger.info("Indexing complete. Store persisted at '%s'.", self.persist_dir)

    def load_existing(self) -> bool:
        store_path = Path(self.persist_dir)
        if not store_path.exists():
            return False
        self._store = Chroma(
            collection_name=self.collection,
            embedding_function=self.embeddings,
            persist_directory=self.persist_dir,
        )
        count = self._store._collection.count()
        if count == 0:
            logger.warning("Chroma store found but contains 0 documents.")
            return False
        logger.info("Loaded existing Chroma store (%d chunks).", count)
        return True

    def get_vector_retriever(self, k: int = CANDIDATE_K):
        if self._store is None:
            raise RuntimeError(
                "Vector store not initialised. "
                "Call add_documents() or load_existing() first."
            )
        return self._store.as_retriever(
            search_type="mmr",
            search_kwargs={"k": k, "fetch_k": k * 3, "lambda_mult": 0.7},
        )

    def get_all_documents(self) -> List[Document]:
        if self._store is None:
            raise RuntimeError("Vector store not initialised.")
        raw = self._store._collection.get(include=["documents", "metadatas"])
        return [
            Document(page_content=text, metadata=meta or {})
            for text, meta in zip(raw["documents"], raw["metadatas"])
        ]

```

```python
# =============================================================================
# STAGE 3 — HYBRID RETRIEVAL  (HybridRetriever)
# =============================================================================

class HybridRetriever:
    """
    EnsembleRetriever combining:
      • MMR Vector Search  (0.6 weight) — conceptual / semantic matches
      • BM25 Keyword Search (0.4 weight) — exact formula / ticker / code matches

    A content-hash deduplication pass removes duplicate chunks that the
    EnsembleRetriever can surface when both tracks return the same passage.
    """

    def __init__(
        self,
        vector_retriever,
        corpus_documents: List[Document],
        vector_weight: float = ENSEMBLE_VECTOR_WEIGHT,
        bm25_weight: float   = ENSEMBLE_BM25_WEIGHT,
        k: int               = CANDIDATE_K,
    ):
        if abs(vector_weight + bm25_weight - 1.0) > 1e-6:
            raise ValueError("Ensemble weights must sum to 1.0")

        logger.info(
            "Initialising BM25 retriever over %d documents…", len(corpus_documents)
        )
        bm25_retriever   = BM25Retriever.from_documents(corpus_documents)
        bm25_retriever.k = k

        self._ensemble = EnsembleRetriever(
            retrievers=[vector_retriever, bm25_retriever],
            weights=[vector_weight, bm25_weight],
        )
        self._k = k
        logger.info(
            "HybridRetriever ready (vector=%.1f, bm25=%.1f, k=%d).",
            vector_weight, bm25_weight, k,
        )

    def retrieve(self, query: str) -> List[Document]:
        results = self._ensemble.invoke(query)
        # Deduplicate by first-200-char content hash
        seen, unique = set(), []
        for doc in results:
            key = hash(doc.page_content[:200])
            if key not in seen:
                seen.add(key)
                unique.append(doc)
        logger.info(
            "HybridRetriever returned %d unique candidate chunks.", len(unique)
        )
        return unique[: self._k]

```

```python
# =============================================================================
# STAGE 4 — CROSS-ENCODER RE-RANKING  (CrossEncoderReranker)
# =============================================================================

class CrossEncoderReranker:
    """
    BAAI/bge-reranker-large performs full cross-attention over each
    (query, passage) pair — far more accurate than cosine similarity alone.

    Compresses 25 candidates → top 5, storing rerank_score and rerank_rank
    back into each document's metadata for downstream citation display.
    """

    def __init__(
        self,
        model_name: str = RERANKER_MODEL_NAME,
        top_k: int      = RERANKED_TOP_K,
    ):
        logger.info(
            "Loading CrossEncoder reranker: %s  (first run downloads ~1.1 GB)",
            model_name,
        )
        self._model = CrossEncoder(model_name, max_length=512)
        self._top_k = top_k
        logger.info("CrossEncoder loaded.")

    def rerank(self, query: str, candidates: List[Document]) -> List[Document]:
        if not candidates:
            return []

        pairs  = [[query, doc.page_content] for doc in candidates]
        scores = self._model.predict(pairs)

        scored = sorted(zip(scores, candidates), key=lambda x: x[0], reverse=True)
        top    = scored[: self._top_k]

        reranked_docs = []
        for rank, (score, doc) in enumerate(top, start=1):
            doc.metadata["rerank_score"] = round(float(score), 4)
            doc.metadata["rerank_rank"]  = rank
            reranked_docs.append(doc)

        logger.info(
            "Re-ranked %d → %d docs. Top score: %.4f",
            len(candidates), len(reranked_docs), float(top[0][0]),
        )
        return reranked_docs

```

```python
# =============================================================================
# STAGE 5 — CONTEXT COMPRESSION & LLM SYNTHESIS  (SynthesisEngine)
# =============================================================================

class SynthesisEngine:
    """
    Two-step synthesis:

    Step A — LLMChainExtractor (Gemini 2.5 Flash, free) strips each of the
             top-5 re-ranked chunks down to only the sentences that answer
             the query, eliminating token waste from irrelevant filler.

    Step B — Gemini 2.5 Flash with the strict anti-hallucination system
             prompt receives the compressed, numbered context and produces
             a cited, structured answer.

    temperature=0.0 on both models guarantees deterministic, factual output.
    """

    def __init__(self, api_key: str):
        self._llm_synthesizer = self._init_gemini(
            api_key=api_key,
            model=GEMINI_SYNTHESIS_MODEL,
            max_tokens=2048,
        )
        self._llm_compressor = self._init_gemini(
            api_key=api_key,
            model=GEMINI_COMPRESSOR_MODEL,
            max_tokens=1024,
        )
        self._compressor = LLMChainExtractor.from_llm(self._llm_compressor)
        self._chain      = self._build_synthesis_chain()

    @staticmethod
    def _init_gemini(api_key: str, model: str, max_tokens: int) -> ChatGoogleGenerativeAI:
        logger.info("Initialising Gemini model: %s", model)
        return ChatGoogleGenerativeAI(
            model=model,
            google_api_key=api_key,
            temperature=0.0,
            max_output_tokens=max_tokens,
            convert_system_message_to_human=False,
        )

    def _build_synthesis_chain(self):
        prompt = ChatPromptTemplate.from_messages([
            ("system", SYSTEM_PROMPT),
            ("human",  HUMAN_PROMPT),
        ])
        return prompt | self._llm_synthesizer | StrOutputParser()

    # ── Context compression ───────────────────────────────────────────────────

    def compress(self, query: str, documents: List[Document]) -> List[Document]:
        """
        Asks LLMChainExtractor to distil each chunk to query-relevant sentences.
        Falls back to the original chunk if the compressor returns nothing.
        """
        compressed: List[Document] = []
        for doc in documents:
            try:
                result = self._compressor.compress_documents([doc], query)
                compressed.extend(result if result else [doc])
            except Exception as exc:
                logger.warning(
                    "Compression failed for a chunk (%s). Keeping original.", exc
                )
                compressed.append(doc)
        logger.info(
            "Compressed %d docs → %d non-empty fragments.",
            len(documents), len(compressed),
        )
        return compressed

    # ── Context formatter ─────────────────────────────────────────────────────

    @staticmethod
    def _format_context(documents: List[Document]) -> str:
        parts = []
        for i, doc in enumerate(documents, start=1):
            m       = doc.metadata
            source  = m.get("source", "unknown")
            page    = m.get("page", m.get("page_number", "?"))
            section = m.get("section_path", "")
            score   = m.get("rerank_score", "?")

            header = f"[Passage {i}] Source: {source} | Page: {page}"
            if section:
                header += f" | Section: {section}"
            header += f" | Relevance Score: {score}"
            parts.append(f"{header}\n{doc.page_content}")

        return "\n\n" + "\n\n---\n\n".join(parts) + "\n"

    # ── Public synthesis entry-point ──────────────────────────────────────────

    def synthesize(
        self, query: str, documents: List[Document]
    ) -> Tuple[str, List[Document]]:
        compressed = self.compress(query, documents)
        context    = self._format_context(compressed)
        logger.info("Invoking Gemini for synthesis…")
        answer = self._chain.invoke({"context": context, "question": query})
        return answer, compressed

```

```python
# =============================================================================
# STAGE 6 — RAGAS EVALUATION FRAMEWORK  (RagasEvaluator)
# =============================================================================

class RagasEvaluator:
    """
    Evaluates pipeline output using four Ragas metrics:

      faithfulness      — claims grounded in retrieved context  (anti-hallucination)
      answer_relevancy  — answer addresses the actual question
      context_recall    — retrieved context covers the ground truth
      context_precision — retrieved context is noise-free

    Uses Gemini 2.5 Flash as the judge LLM and Google's embedding-001 as
    the embedding backbone — BOTH are free via the same GOOGLE_API_KEY.
    No OpenAI key is needed anywhere in this pipeline.
    """

    METRICS = [faithfulness, answer_relevancy, context_recall, context_precision]

    def __init__(self, api_key: str):
        self._api_key = api_key
        self._judge_llm, self._judge_embeddings = self._init_ragas_judge()

    def _init_ragas_judge(self):
        logger.info(
            "Initialising Ragas judge LLM & embeddings using Gemini (free tier)…"
        )
        judge_llm = LangchainLLMWrapper(
            ChatGoogleGenerativeAI(
                model=GEMINI_RAGAS_MODEL,
                google_api_key=self._api_key,
                temperature=0.0,
                max_output_tokens=1024,
                convert_system_message_to_human=False,
            )
        )
        judge_embeddings = LangchainEmbeddingsWrapper(
            GoogleGenerativeAIEmbeddings(
                model=GEMINI_RAGAS_EMBED_MODEL,
                google_api_key=self._api_key,
                task_type="retrieval_document",
            )
        )
        return judge_llm, judge_embeddings

    def evaluate(self, samples: List[dict]) -> pd.DataFrame:
        """
        samples: list of dicts, each with keys:
          question (str), answer (str), contexts (List[str]), ground_truth (str)
        Returns a Pandas DataFrame with per-sample and aggregate scores.
        """
        logger.info("Running Ragas evaluation on %d sample(s)…", len(samples))

        dataset = Dataset.from_dict({
            "question":    [s["question"]     for s in samples],
            "answer":      [s["answer"]       for s in samples],
            "contexts":    [s["contexts"]     for s in samples],
            "ground_truth":[s["ground_truth"] for s in samples],
        })

        # Wire Gemini as the judge — overrides Ragas' default OpenAI judge
        for metric in self.METRICS:
            metric.llm        = self._judge_llm
            metric.embeddings = self._judge_embeddings

        results = evaluate(
            dataset=dataset,
            metrics=self.METRICS,
            llm=self._judge_llm,
            embeddings=self._judge_embeddings,
        )

        df = results.to_pandas()
        logger.info("Ragas evaluation complete.")
        return df
```

```python
# =============================================================================
# ORCHESTRATION CLASS  (AdvancedRAGPipeline)
# =============================================================================

class AdvancedRAGPipeline:
    """
    Composes all six pipeline stages into one object.

    Public API:
      pipeline.ingest(file_path)          — load, chunk, embed, index
      pipeline.query(question) → dict     — full retrieval → synthesis
      pipeline.evaluate(samples) → DataFrame — Ragas scorecard
    """

    def __init__(self):
        logger.info("=" * 70)
        logger.info("Initialising AdvancedRAGPipeline  (100%% free — Gemini API)")
        logger.info("=" * 70)

        # Resolve & validate API key once at startup
        self._api_key = _get_google_api_key()

        # Instantiate all stages
        self._ingestion_engine = DocumentIngestionEngine()
        self._vector_manager   = VectorStoreManager()
        self._reranker         = CrossEncoderReranker()
        self._synthesis_engine = SynthesisEngine(api_key=self._api_key)
        self._evaluator        = RagasEvaluator(api_key=self._api_key)

        # Hybrid retriever is built after at least one document is ingested
        self._hybrid_retriever: Optional[HybridRetriever] = None

        # Try to restore a previously persisted store
        if self._vector_manager.load_existing():
            self._rebuild_hybrid_retriever()

        logger.info("AdvancedRAGPipeline ready.")

    # ── Ingestion ─────────────────────────────────────────────────────────────

    def ingest(self, file_path: str) -> None:
        logger.info("─" * 60)
        logger.info("INGESTION: %s", file_path)
        logger.info("─" * 60)

        chunks = self._ingestion_engine.ingest(file_path)   # Stage 1
        self._vector_manager.add_documents(chunks)           # Stage 2
        self._rebuild_hybrid_retriever()                     # Stage 3 setup

        logger.info("Ingestion complete for '%s'.", file_path)

    def _rebuild_hybrid_retriever(self) -> None:
        all_docs    = self._vector_manager.get_all_documents()
        vector_ret  = self._vector_manager.get_vector_retriever(k=CANDIDATE_K)
        self._hybrid_retriever = HybridRetriever(
            vector_retriever=vector_ret,
            corpus_documents=all_docs,
            k=CANDIDATE_K,
        )

    # ── Query ─────────────────────────────────────────────────────────────────

    def query(self, question: str) -> dict:
        """
        Runs the full pipeline:
          Stage 3 — Hybrid Retrieval   (25 candidates)
          Stage 4 — CrossEncoder Rerank (top 5)
          Stage 5 — Compression + Gemini synthesis

        Returns dict with keys:
          question, answer, source_documents, contexts
        """
        if self._hybrid_retriever is None:
            raise RuntimeError(
                "No documents ingested yet. Call pipeline.ingest(file_path) first."
            )

        logger.info("─" * 60)
        logger.info("QUERY: %s", question)
        logger.info("─" * 60)

        candidates = self._hybrid_retriever.retrieve(question)   # Stage 3
        reranked   = self._reranker.rerank(question, candidates)  # Stage 4
        answer, compressed_docs = self._synthesis_engine.synthesize(
            question, reranked                                     # Stage 5
        )

        logger.info("Query complete.")
        return {
            "question":         question,
            "answer":           answer,
            "source_documents": compressed_docs,
            "contexts":         [doc.page_content for doc in compressed_docs],
        }

    # ── Evaluation ────────────────────────────────────────────────────────────

    def evaluate(self, samples: List[dict]) -> pd.DataFrame:
        """Stage 6 — Ragas scorecard using Gemini as judge."""
        return self._evaluator.evaluate(samples)
```

```python
@staticmethod
def _init_embeddings() -> HuggingFaceEmbeddings:
    logger.info(
        "Loading embedding model: %s  (first run downloads ~1.2 GB)",
        EMBEDDING_MODEL_NAME,
    )
    # BGE models need a query-time instruction prefix for retrieval tasks.
    # 'query_instruction' was removed from HuggingFaceEmbeddings in langchain-huggingface >= 0.1.
    # The correct way is to pass it via encode_kwargs using the sentence-transformers
    # 'prompt' key, which prepends the instruction at encode time without storing it in the index.
    return HuggingFaceEmbeddings(
        model_name=EMBEDDING_MODEL_NAME,
        model_kwargs={"device": "cpu"},        # change to "cuda" if GPU is available
        encode_kwargs={
            "normalize_embeddings": True,
            "prompt": "Represent this sentence for searching relevant passages: ",
        },
    )
```
